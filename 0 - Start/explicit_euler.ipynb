# Explicit Euler
Solving an ODE with Explicit Euler

```python
import numpy as np
```

```python
k1 = 1.0 # Param 1
k2 = 0.5 # Param 2
k3 = 0.2 # Param 3

A_INIT = 10.0
B_INIT = 0.0
# Initial values of variables A and B
```

```python
def dash (AB: np.ndarray): 
    return np.matmul([
        [-k1, k2], 
        [k1, -(k2+k3)]], 
    AB)
```

```python
def simulate(timesteps: int, delta = 0.001):
    AB = np.array([A_INIT, B_INIT])
    sim = [AB.copy()]
    for _ in range(timesteps):
        AB = AB + (delta * dash(AB))
        sim.append(AB.copy())
        
    return sim
```

```python
import matplotlib.pyplot as plt

plt.plot(simulate(10000))
plt.legend(["A", "B"])
plt.show()
```
