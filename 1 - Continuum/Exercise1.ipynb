## Q2 - Predator-Prey System Analysis

$dX = c_2[X][Y] + r_1[X](1 - \frac{[X]}{K})$

$dY = c_1[X][Y] + c_2([X][Y])^2 - r_2[Y]$

```python
r1 = 0.000001
r2 = 0.00002
K  = 10
c1 = 0.09
c2 = 0.05

X_INIT = 10
Y_INIT = 10
```

```python
X_log = [X_INIT]
Y_log = [Y_INIT]

DELTA = 0.001

def dash(X, Y):
    dX = (c2*X*Y) + (r1*X*(1- (X/K)))
    dY = (c1*X*Y) + (c2*(X*Y)**2) - (r2*Y)
    return dX, dY

for _ in range(1000):
    dX, dY = dash(X_log[-1], Y_log[-1])
    X = X_log[-1] + (DELTA * dX)
    Y = Y_log[-1] + (DELTA * dY)
    X_log.append(X)
    Y_log.append(Y)
    
```

```python
import matplotlib.pyplot as plt

plt.plot(X_log, Y_log)
plt.show()
```
